```python
import pandas as pd

# المسار الدقيق للملف
file_path = '/kaggle/input/datasets/mkechinov/ecommerce-behavior-data-from-multi-category-store/2019-Oct.csv'

# قراءة عينة بحجم 500,000 صف
df = pd.read_csv(file_path, nrows=500000)

# 1. معاينة أولية (خذ سكرين شوت هنا لـ Task 1)
print("--- Task 1: Exploration ---")
print("Data Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
print("\nFirst 5 rows:")
display(df.head())

# 2. التنظيف ومعالجة القيم المفقودة (Task 2)
print("\n--- Task 2: Cleaning & Transformation ---")

# إزالة التكرارات
df.drop_duplicates(inplace=True)

# معالجة القيم الفارغة
df['brand'] = df['brand'].fillna('Unknown')
df['category_code'] = df['category_code'].fillna('unknown.unknown')

# استخراج الفئة الرئيسية (Main Category)
df['main_category'] = df['category_code'].apply(lambda x: str(x).split('.')[0])

# تحويل الوقت واستخراج التاريخ
df['event_time'] = pd.to_datetime(df['event_time'])
df['event_date'] = df['event_time'].dt.date

# تصفية الأسعار غير المنطقية
df = df[df['price'] > 0]

print("Shape after cleaning:", df.shape)

# 3. تصدير الملف بصيغة CSV لتنزيله لـ Power BI
output_file = 'ecommerce_cleaned_sample.csv'
df.to_csv(output_file, index=False)
print(f"\nتم حفظ الملف بنجاح كـ: {output_file}")
```
